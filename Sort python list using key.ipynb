Sort Python List using Key

```python
people = [ 
    {"Name": "Alice", "Age" : 28},
    {"Name": "Bob", "Age" : 35},
    {"Name": "Charlie", "Age" : 22}
]

sorted_people = sorted(people, key=lambda x: x['Age'])
print(sorted_people)

names = ["Alice", "Bob", "Charlie"]
sorted_names = sorted(names, key= len)
print(sorted_names)
```

Filter Data in Python

```python
# filter (function, iterable)

data = [100,200,300,400]

filtered = list(filter(lambda x: x > 150, data))
print(filtered)
```

```python
data = [100,200,300,400]

filtered = list(filter(lambda x: x > 250, data))
print(filtered)
```

```python
data = [100, -1, 200, None, 0]

cleared_data = list(filter(lambda x:x is not None and x > 0, data)) 
print(cleared_data)
```

```python
employees = [
    {"name": "Alice", "Salary": 70000},
    {"name": "Bob", "Salary": 50000},
    {"name": "Charlie", "Salary": 60000}
]

high_earners = list(filter(lambda emp: emp["Salary"] > 55000, employees))
print(high_earners)

indians = list(filter(lambda x: x["Salary"] == 70000, employees))
print(indians)
```

Set() function in Python
- removing duplicates
- performing set operations (union, intersection, etc)

```python
name = ['Chintan', 'Tutedude', 'Data', 'Engineering']
unique_name = list(set(name))
print(unique_name)
```

```python
# Non common elements in name and name1

name = {'Chintan', 'Tutedude', 'Data', 'Engineering'}
name1 = {'Chintan', 'Tutedude'}

non_common = name.difference(name1)
print(non_common)
```

```python
datra = [
    {"name": "Alice", "Department" : "HR"},
    {"name": "Bob", "Department" : "IT"},
    {"name": "Charlie", "Department" : "Finance"},
    {"name": "David", "Department" : "IT"}
]

unique_values = set(map(lambda x: x['Department'], datra))
print(unique_values)
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['Month'] = df['order_date'].dt.to_period('M')
count = df.groupby(['Month', 'order_status'])['order_id'].count().reset_index()
print(count)
```

```python
pivot = df.pivot_table(index='Month', columns='order_status', values='order_id', aggfunc='count')
print(pivot)
```

Joining Data Using Pandas

```python
def get_column_names(schemas, ds_name, sorting_key = 'column_position'):
    column_details = schemas[ds_name]
    columns = sorted(column_details, key=lambda x: x[sorting_key])
    return [column['column_name'] for column in columns]
```

```python

import json

schemas = json.load(open('schemas.json'))
```

```python
order_column = get_column_names(schemas, 'orders')
```

```python
orders = pd.read_csv('Ordersss.txt', names=order_column)
```

```python
customers_column = get_column_names(schemas, 'customers')
```

```python
customers_column
```

```python
customers = pd.read_csv('Ordersss.txt', names = customers_column)
```
